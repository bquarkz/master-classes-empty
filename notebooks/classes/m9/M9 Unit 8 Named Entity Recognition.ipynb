# Named Entity Recognition Demonstration In Python

## Background

In today’s digital world, a vast amount of information is generated in the form of text, such as emails, news articles, social media posts, and online content. While this text is easy for humans to read and understand, it is considered unstructured data, making it challenging for computers to process directly.

To make sense of such data, we use techniques from Natural Language Processing (NLP). One important task in NLP is Named Entity Recognition (NER), which helps in identifying and classifying key elements in text—such as names of people, places, organizations, and dates.

By extracting these entities, NER enables machines to better understand the context of the text and convert unstructured information into a more structured and usable for

## Objective
To demonstrate how Named Entity Recognition (NER) can automatically identify and extract key information such as names, places, organizations, and dates from unstructured text using Python.

### Import Libraries

```python
import spacy
from spacy import displacy
```

### Load the Model

```python
# Load English model
nlp = spacy.load("en_core_web_sm")
```

### Text Processing

```python
text = "Elon Musk announced that Tesla will expand its operations in Texas and collaborate with Meta on AI research starting in September."
# Process the text
doc = nlp(text)
```

### Output

```python
# Print named entities
print("Named Entities:\n")
for ent in doc.ents:
    print(f"{ent.text} --> {ent.label_}")
    
displacy.render(doc, style="ent", jupyter=True)
```

```python
text = "John invested $5000 in Apple stocks and plans to sell them in December after the earnings report."
doc = nlp(text)

print("Named Entities:\n")
for ent in doc.ents:
    print(f"{ent.text} --> {ent.label_}")

displacy.render(doc, style="ent", jupyter=True)
```
